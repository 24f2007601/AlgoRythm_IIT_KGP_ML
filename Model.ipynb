```python
import pandas as pd
from pycaret.regression import *
```

```python
train_df = pd.read_csv('datasets/train_dataset.csv')

 
reg_setup = setup(
    data=train_df,
    target='overall_yield',
    session_id=123,
    train_size=0.8,
    normalize=True,
    fold=10
)
 
best_model = compare_models()
final_model = finalize_model(best_model)
save_model(final_model, 'best_yield_model')
```

```python
test_df = pd.read_csv('datasets/test_dataset.csv')
print("Test shape:", test_df.shape)
 
new_predictions = predict_model(final_model, data=test_df)
print(new_predictions.head())
```

```python
new_predictions.to_csv('test_predictions.csv', index=False)
print("Saved predictions to test_predictions.csv")
```
